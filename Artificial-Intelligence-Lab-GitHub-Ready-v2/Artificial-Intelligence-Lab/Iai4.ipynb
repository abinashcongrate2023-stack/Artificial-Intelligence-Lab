```python
graph = {
    'A': [('B', 1), ('C', 3)],
    'B': [('D', 2), ('E', 6)],
    'C': [('F', 5)],
    'D': [('G', 2)],
    'E': [('G', 1)],
    'F': [('G', 2)],
    'G': []
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 4,
    'D': 2,
    'E': 1,
    'F': 2,
    'G': 0,
}

def astar(start, goal):
    open_list = [(start, 0)]
    came_from = {}
    g_cost = {start: 0}

    while open_list:
        # Find the node with the lowest f_cost (g_cost + heuristic)
        current = min(open_list, key=lambda x: x[1] + heuristic[x[0]])
        current_node = current[0]

        # FIX 1: Remove the evaluated node from open_list to avoid infinite loops
        open_list.remove(current)

        # FIX 2: Corrected typo 'cureent_node' to 'current_node'
        if current_node == goal:
            path = [goal]
            while current_node in came_from:
                current_node = came_from[current_node]
                path.append(current_node)
            path.reverse()
            return path, g_cost[goal]

        for neighbor, cost in graph[current_node]:
            new_cost = g_cost[current_node] + cost
            if neighbor not in g_cost or new_cost < g_cost[neighbor]:
                g_cost[neighbor] = new_cost
                came_from[neighbor] = current_node
                open_list.append((neighbor, new_cost))

    # FIX 3: Unindented this line so it triggers only after open_list is completely empty
    return None, float('inf')

# FIX 4: Added missing parentheses () to execute the .upper() function
start = input("Enter starting node: ").upper()
goal = input("Enter goal node: ").upper()

if start not in graph or goal not in graph:
    print("Invalid node")
else:
    path, cost = astar(start, goal)

if path:
    print("\nShortest path:", "->".join(path))
    print("Total cost:", cost)
else:
    print("No path found")
```
