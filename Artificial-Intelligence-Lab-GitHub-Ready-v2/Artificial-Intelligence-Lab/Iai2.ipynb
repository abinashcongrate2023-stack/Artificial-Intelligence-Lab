```python
# Function to perform DFS
def dfs(graph, node, visited):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)
        # Check if node exists in graph to prevent KeyError
        if node in graph:
            for neighbour in graph[node]:
                dfs(graph, neighbour, visited)

graph = {}
n = int(input("Enter the number of vertices: "))

for i in range(n):
    vertex = input(f"Enter vertex {i+1}: ")
    neighbour = input(f"Enter adjacent vertices of {vertex} (space-separated): ").split()
    graph[vertex] = neighbour

start = input("Enter the starting vertex: ")

print("\nGraph:", graph)
print("DFS Traversal:", end=" ")
visited = set()
dfs(graph, start, visited)
print()  # For a clean newline at the end
```
