```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    print("BFS Traversal: ", end="")
    while queue:
        vertex = queue.popleft()  # Corrected spelling from vertax
        print(vertex, end=" ")    # Corrected spelling from vertex

        for neighbor in graph[vertex]:  # Corrected spelling from vertax
            if neighbor not in visited:
                visited.add(neighbor)   # Corrected spelling from nighbor
                queue.append(neighbor)

graph = {}
n = int(input("Enter the number of vertices: "))

for i in range(n):
    vertex = input(f"Enter vertex {i+1}: ")
    neighbors = input(f"Enter adjacent vertices of {vertex} (space-separated): ").split()  # Corrected spelling from nighbors
    graph[vertex] = neighbors

start = input("Enter the starting vertex: ")
print("\nGraph:", graph)
bfs(graph, start)
```
